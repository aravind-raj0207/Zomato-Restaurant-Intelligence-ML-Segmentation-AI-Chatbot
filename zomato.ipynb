```python
df["rating_status"] = np.where(
    df["aggregate_rating"] == 0,
    "Not Rated",
    "Rated"
)
```

```python
df["rating_clean"] = df["aggregate_rating"].replace(0, np.nan)
```

```python
print(df[["aggregate_rating", "rating_clean", "rating_status"]].head(10))
```

```python
binary_cols = [
    "has_table_booking",
    "has_online_delivery",
    "is_delivering_now"
]

for col in binary_cols:
    df[col] = df[col].map({
        "Yes": 1,
        "No": 0
    })
```

```python
df.head()
```

```python
df.drop(columns=["switch_to_order_menu"], inplace=True)
```

```python
df["log_votes"] = np.log1p(df["votes"])
df["log_votes"]
```

```python
#easy to create dashboard
price_map = {
    1: "Budget",
    2: "Mid-Range",
    3: "Premium",
    4: "Luxury"
}

df["price_category"] = df["price_range"].map(price_map)
```

```python
df.head()
```

```python
def rating_category(rating):
    if pd.isna(rating):
        return "Not Rated"
    elif rating < 2.5:
        return "Poor"
    elif rating < 3.5:
        return "Average"
    elif rating < 4.0:
        return "Good"
    elif rating < 4.5:
        return "Very Good"
    else:
        return "Excellent"

df["rating_category"] = df["rating_clean"].apply(rating_category)
```

```python
df.head()
```

```python
#creating primary cuisines
df["primary_cuisine"] = (
    df["cuisines"]
    .fillna("Unknown")
    .str.split(",")
    .str[0]
    .str.strip()
)
```
